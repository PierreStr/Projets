# Analyse de la qualité des jeux de données

### Créer une fonction qui produit les informations suivantes pour un data frame :
- la taille,
- le typage,
- le pourcentage de null.

```python
## Rediriger la racine du notebooks au dossier Projet_perso ###
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
sys.path.append(str(PROJECT_ROOT))
#############################################################

import pandas as pd
from python_scripts.DB_link import engine
```

```python
def quality_report(df, col_min_max, col_mode, col_mean_med):
    """
    Génère un rapport de qualité d'un DataFrame.
    """
    if len(col_mode) > 0:
        report = pd.DataFrame({
            "Colonne": df.columns,
            "Type": df.dtypes.astype(str).values,
            "Non nuls": df.notna().sum().values,
            "% Null": (df.isna().mean() * 100).round(2).values,
            "Distincts": df.nunique(dropna=True).values,
            "% Distincts": ((df.nunique(dropna=True) / len(df)) * 100).round(2).values,
            "Mode": [
                df[col].mode().iloc[0] if col in col_mode else "-"
                for col in df.columns
            ]
        })
    else:
        report = pd.DataFrame({
            "Colonne": df.columns,
            "Type": df.dtypes.astype(str).values,
            "Non nuls": df.notna().sum().values,
            "% Null": (df.isna().mean() * 100).round(2).values,
            "Distincts": df.nunique(dropna=True).values,
            "% Distincts": ((df.nunique(dropna=True) / len(df)) * 100).round(2).values,
        })

    if len(col_min_max) > 0:
        # Colonnes numériques
        report["Min"] = [df[col].min() if col in col_min_max else "-" for col in df.columns]

        report["Max"] = [
            df[col].max() if col in col_min_max else "-"
            for col in df.columns
        ]
    if len(col_mean_med) > 0:
        report["Moyenne"] = [
            round(df[col].mean(), 2) if col in col_mean_med else "-"
            for col in df.columns
        ]

        report["Médiane"] = [
            round(df[col].median(), 2) if col in col_mean_med else "-"
            for col in df.columns
        ]

    return report
```

```python
accidents = pd.read_sql_table("accidents", schema="silver", con=engine)

columns_for_mode = [
    'GEO_region', 'GEO_province', 'GEO_commune', 'COND_carrefour', 'COND_meteo', 'COND_route', 'COND_agglomeration',
    'COND_lumiere', 'COND_type_route', 'ACC_etat_victimes','ACC_vehicule_1', 'ACC_vehicule_2', 'ACC_collision_type',
    'COND_obstacles', 'ACC_nom_autoroute', 'COND_routes_largeur_cat'
]

columns_for_min_max = [
    'COND_densite', 'SOIN_min_distance'
]

quality_report(accidents, columns_for_min_max, columns_for_mode, columns_for_min_max)
```

```python
population = {}
for year in range(2017, 2025):
    population[f"population_{year}"] = pd.read_sql_table(f"population_{year}", schema="silver", con=engine)
```

```python
columns_for_mode = []

for year in range(2017, 2025):
    print(f"Report for data frame population ({year})")
    display(quality_report(population[f"population_{year}"], [], columns_for_mode, []))
```

```python
occupation_sol = pd.read_sql_table("occupation_par_commune", schema="silver", con=engine)

columns_for_min_max = ['OCC_% Autres', 'Eaux',
       'Total', 'OCC_% Total terrains bâtis', 'OCC_% Autres terrains bâtis',
       'Terrains commerciaux', 'Terrains industriels', 'Terrains résidentiels',
       'OCC_% Terrains agricoles', 'OCC_% Terrains forestiers']

quality_report(occupation_sol, columns_for_min_max, [], columns_for_min_max)
```

```python
revenus = {}
for year in range(2017, 2024):
    revenus[f"revenus_{year}"] = pd.read_sql_table(f"revenus_{year}", schema="silver", con=engine)

columns_for_mode = []

columns_for_min_max = ['REV_pourcentage_manquant', 'REV_median', 'REV_risque_pauvrete']

for year in range(2017, 2024):
    print(f"Report for data rame revenus ({year})")
    display(quality_report(revenus[f"revenus_{year}"], columns_for_min_max, columns_for_mode, columns_for_min_max))
```
